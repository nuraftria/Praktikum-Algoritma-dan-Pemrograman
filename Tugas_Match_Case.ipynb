#Pengenalan match case vs if-elif-else

```python
#Contoh Logika dengan if-elif-else Tradisional
status_code=404

if status_code==200:
  print("OK:Permintaan berhasil.")
elif status_code==400:
  print("Bad Request:Permintaan tidak valid.")
elif status_code==404:
  print("Not Found:Halaman tidak ditemukan.")
elif status_code==500:
  print("Internal Server Error:Kesalahan server.")
else:
  print("Status Code tidak dikenal.")
```

```python
#Contoh Logika setara menggunakan match-case
status_code=404

match status_code:
  case 200:
    print("OK:Permintaan berhasil.")
  case 400:
    print("Bad Request:Permintaan tidak valid.")
  case 404:
    print("Not Found:Halaman tidak ditemukan.")
  case 500:
    print("Internal Server Error:Kesalahan server.")
  case _:#Wilcard '_' berfungsi sebagai default/else:
    print("Status Code tidak dikenal.")
```

#Pencocokan Banyak Nilai dengan Operator OR (|)

```python
def cek_kategori_hari(hari):
  match hari.title():
    case "Senin"|"Selasa"|"Rabu"|"Kamis"|"Jumat":
      return f"{hari}adalah Hari Kerja (Weekday)."
    case "Sabtu"|"Minggu":
      return f"{hari}adalah Akhir Pekan (Weekend)."
    case _:
      return "Nama hari tidak valid!"

#Uji Coba Fungsi
print(cek_kategori_hari("senin"))
print(cek_kategori_hari("Sabtu"))
print(cek_kategori_hari("Kiamat"))
```

#Menambahkan Syarat Logika dengan Guard Clauses (if dalam case)

```python
def analisis_angka(angka):
  match angka:
    case x if x>0 and x%2==0:
      print(f"{angka} adalah Bilangan Positif Genap.")
    case x if x>0 and x%2!=0:
      print(f"{angka} adalah Bilangan Positif Ganjil.")
    case x if x<0:
      print(f"{angka} adalah Bilangan Negatif.")
    case 0:
      print("Angka adalah Nol.")
    case _:
      print("Input bukan angka valid.")

#Uji Coba
analisis_angka(14)
analisis_angka(7)
analisis_angka(-5)
analisis_angka(0)
```

#Pattering Matching pada Struktur Data(Sequence dan Dictionary)

#A. Pencocokan pada List/Tuple(Sequence Matching)

```python
def eksekusi_perintah(command):
  match command:
    case ["move",arah]:
      print(f"Karakter bergerak ke arah: {arah}")
    case ["attack", senjata, target]:
      print(f"Menyerang {target} menggunakan {senjata}!")
    case ["quit"] | ["exit"]:
      print("Keluar dari permainan...")
    case _:
      print("Perintah tidak dikenal!")

#Uji Coba Perintah Game
eksekusi_perintah(["move", "Utara"])
eksekusi_perintah(["attack", "Pedang", "Goblin"])
eksekusi_perintah(["exit"])
```

#B. Pencocokan pada Dictionary

```python
def proses_pengguna(user_data):
  match user_data:
    case {"nama": nama, "role": "admin"}:
      print(f"Selamat datang Admin {nama}. Akses Penuh Diberikan.")
    case {"nama": nama, "role": "user", "aktif": True}:
      print(f"Halo Pengguna {nama}, akun Anda aktif")
    case {"nama": nama, "role": "user", "aktif": False}:
      print(f"Peringatan: Akun {nama} dalam status non-aktif!")
    case _:
      print("Format data pengguna tidak valid!")

#Uji Coba
proses_pengguna({"nama": "Budi", "role": "admin"})
proses_pengguna({"nama": "Siti", "role": "user", "aktif": True})
proses_pengguna({"nama": "Andi", "role": "user", "aktif": False})
```

#Latihan 1: Sistem Kasir Sederhana (Diskon Berdasarkan Hari)

Buatlah fungsi hitung_diskon(total_belanja, hari):

  1. Hari Jumat atau Sabtu: diskon 15% jika total_belanja >= 100.000, selain itu diskon 5%.
  2. Hari Minggu: diskon 10% untuk semua total_belanja.
  3. Hari lainnya: tidak ada diskon (0%). Gunakan match-case dan guard clause!



```python
#Lengkapi kode fungsi di bawah ini
def hitung_diskon(total_belanja, hari):
  match hari.title():
    case "Jumat" | "Sabtu" if total_belanja >= 100000:
      diskon = 0.15
    case "Jumat" | "Sabtu":
      diskon = 0.05
    case "Minggu":
      diskon = 0.10
    case _:
      diskon = 0.0

  potongan = total_belanja * diskon
  total_bayar = total_belanja - potongan
  return total_bayar

#Uji coba kasir
print("Total bayar (Jumat, 150rb):", hitung_diskon(150000, "jumat"))
print("Total bayar (Minggu, 50rb):", hitung_diskon(50000, "minggu"))
print("Total bayar (Senin, 200rb):", hitung_diskon(200000, "senin"))
```

### Soal 1: Parser Respons API Cuaca
Buatlah sebuah fungsi `parse_cuaca_response(response)` yang menerima masukan berupa Dictionary respons API cuaca.

Aturan Pencocokan:
1. Jika dictionary memiliki bentuk `{"status": 200, "kota": kota, "suhu": suhu}`:
   - Cetak: `"Cuaca di <kota> saat ini adalah <suhu>°C."`
2. Jika dictionary memiliki bentuk `{"status": 404, "pesan": pesan}`:
   - Cetak: `"Gagal mendapatkan data: <pesan>"`
3. Jika dictionary memiliki bentuk `{"status": 500}`:
   - Cetak: `"Layanan API Cuaca sedang gangguan/down."`
4. Untuk format lainnya:
   - Cetak: `"Format respons tidak dikenal."`

```python
def parse_cuaca_response(response):
  match response:
    case {"status": 200, "kota": kota, "suhu": suhu}:
      print(f"Cuaca di {kota} saai ini adalah {suhu}℃.")
    case {"status": 404, "pesan": pesan}:
      print(f"Gagal mendapatkan data: {pesan}.")
    case {"status": 500}:
      print("Layanan API Cuaca sedang gangguan/down.")
    case _:
      print("Format respons tidak dikenal.")

#Uji Coba
parse_cuaca_response({"status": 200, "kota": "Semarang", "suhu": "32"})
parse_cuaca_response({"status": 404, "pesan": "Kota tidak ditemukan"})
parse_cuaca_response({"status": 500})
```

### Soal 2: Evaluator Nilai & Predikat Akademik
Buatlah fungsi `evaluasi_mahasiswa(data)` yang menerima tuple berisi `(nama, nilai_akhir)`.

Aturan dengan `match-case` & *Guard Clauses*:
1. Nilai >= 85: Cetak `"<nama> LULUS dengan Predikat A (Sangat Memuaskan)"`
2. Nilai >= 70 dan < 85: Cetak `"<nama> LULUS dengan Predikat B (Memuaskan)"`
3. Nilai >= 60 dan < 70: Cetak `"<nama> LULUS dengan Predikat C (Cukup)"`
4. Nilai < 60: Cetak `"<nama> TIDAK LULUS (Perlu Remidial)"`
5. Jika format tuple salah atau nilai di luar rentang 0-100: Cetak `"Data mahasiswa tidak valid!"`

```python
def evaluasi_mahasiswa(data):
  match data:
    case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 85 <= nilai_akhir <= 100:
      print(f"{nama} LULUS dengan Predikat A (Sangat Memuaskan)")
    case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 70 <= nilai_akhir < 85:
      print(f"{nama} LULUS dengan Predikat B (Memuaskan)")
    case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 60 <= nilai_akhir < 70:
      print(f"{nama} TIDAK LULUS dengan Predikat C (Cukup)")
    case (nama, nilai_akhir) if isinstance(nilai_akhir, (int, float)) and 0 <= nilai_akhir < 60:
      print(f"{nama} TIDAK LULUS (Perlu Remedial)")
    case _:
      print("Data mahasiswa tidak valid!")

evaluasi_mahasiswa(("Rian", 88))
evaluasi_mahasiswa(("Dewi", 65))
evaluasi_mahasiswa(("Eko", 45))
evaluasi_mahasiswa(("Budi", 105))
```
